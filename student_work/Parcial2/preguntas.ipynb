R-7.1 Proporcione un algoritmo para encontrar el penúltimo nodo en una lista enlazada simple, en la cual el último nodo se identifica porque su referencia `next` es `None`.

```python
def penultimo(L):
    if L.head is None or L.head.next is None:
        raise ValueError("Lista con menos de 2 elementos")
    else: 
        actual = L.head
        while actual._next._next is not None:
            actual = actual._next
        return actual._next
```

R-7.2 Describe un buen algoritmo para concatenar dos listas simplemente enlazadas(Solo ._next, la “doblemente enlazada” tiene dos ganchos: _prev y _next.), L y M —contando únicamente con referencias al primer nodo de cada una—(No tienes _tail ni _size), de modo que se obtenga una única lista L que contenga todos los nodos de L seguidos de todos los nodos de M.

```python
def concatenar(L, M):
    # Caso borde: si L está vacía, L empieza donde empieza M
    if L._head is None:
        L._head = M._head
    else:
        # Caminar hasta el ÚLTIMO nodo de L (el que tiene _next = None)
        actual = L._head
        while actual._next is not None:
            actual = actual._next
        # Enganchar el último de L con el primero de M
        actual._next = M._head

# TRAZA con L: A → B → C,  M: D → E
# actual=A, _next=B → no es None, avanzar
# actual=B, _next=C → no es None, avanzar
# actual=C, _next=None → parar; C._next = D
# Resultado: A → B → C → D → E
```

Complejidad: O(n), donde n es el tamaño de L, porque hay que recorrerla para encontrar su último nodo. El tamaño de M no importa: nunca la recorres.

```python
def contar(nodo):
    # Caso base: llegamos al final de la lista
    if nodo is None:
        return 0
    else:
        # Cuento este nodo (1) + los nodos que quedan desde el siguiente
        return 1 + contar(nodo._next)

def contar_lista(L):
    # La recursión trabaja con NODOS, así que arrancamos desde el primero
    return contar(L._head)

# TRAZA con A → B → C → None:
# contar(A) = 1 + contar(B)
# contar(B) = 1 + contar(C)
# contar(C) = 1 + contar(None)
# contar(None) = 0  ← caso base, ahora se devuelve hacia arriba
# contar(C) = 1 + 0 = 1 → contar(B) = 1 + 1 = 2 → contar(A) = 1 + 2 = 3
```

C-7.24 Proporciona una implementación completa del TDA pila (stack) utilizando una lista simplemente enlazada que incluya un nodo centinela de cabecera.

```python
# --- Error personalizado (opcional; también podrías usar IndexError) ---
class Empty(Exception):                    # creo un tipo de error llamado "Empty"
    pass                                   # no necesita nada más, solo el nombre


# --- La pila ---
class LinkedStack:                         # la clase de la pila
    """Pila LIFO sobre lista enlazada simple con nodo centinela."""

    class _Node:                           # molde para crear nodos (vagones)
        __slots__ = '_element', '_next'    # cada nodo tiene solo 2 cosas: dato y gancho
        def __init__(self, e, n):          # al crear un nodo le paso el dato (e) y a quién apunta (n)
            self._element = e              # guardo el dato
            self._next = n                 # guardo a qué nodo apunta

    def __init__(self):                    # se ejecuta al crear la pila: LinkedStack()
        self._header = self._Node(None, None)  # creo el centinela: sin dato y sin apuntar a nadie
        self._size = 0                     # la pila empieza con 0 elementos

    def __len__(self):                     # se ejecuta cuando escribo len(pila)
        return self._size                  # devuelvo cuántos elementos hay

    def is_empty(self):                    # pregunta: ¿la pila está vacía?
        return len(self) == 0              # True si hay 0 elementos, False si no

    def push(self, e):                     # meter el dato e encima de la pila
        nuevo = self._Node(e, self._header._next)  # creo un nodo con e que apunta a la cima actual
        self._header._next = nuevo         # el centinela ahora apunta al nodo nuevo (nueva cima)
        self._size += 1                    # hay un elemento más

    def top(self):                         # mirar el dato de arriba SIN quitarlo
        if self.is_empty():                # si no hay nada...
            raise Empty('la pila está vacía')  # ...lanzo error
        else:                              # si hay algo...
            return self._header._next._element  # devuelvo el dato del nodo al que apunta el centinela

    def pop(self):                         # quitar el dato de arriba y devolverlo
        if self.is_empty():                # si no hay nada...
            raise Empty('la pila está vacía')  # ...lanzo error
        else:                              # si hay algo...
            cima = self._header._next      # guardo el nodo de arriba para no perderlo
            self._header._next = cima._next  # el centinela ahora apunta al de abajo (se salta la cima)
            self._size -= 1                # hay un elemento menos
            return cima._element           # devuelvo el dato del nodo que saqué


# --- Tests ---
ls = LinkedStack()                         # pila vacía:  header → None
ls.push(1)                                 # header → 1 → None
ls.push(2)                                 # header → 2 → 1 → None
ls.push(3)                                 # header → 3 → 2 → 1 → None
ls.push(4)                                 # header → 4 → 3 → 2 → 1 → None

print(len(ls), ls.top())                   # 4  4   (top solo mira, no quita)
print(ls.pop())                            # 4      queda: header → 3 → 2 → 1
print(ls.pop())                            # 3      queda: header → 2 → 1
print(len(ls))                             # 2
```

C-7.25 Proporciona una implementación completa del TDA cola (queue ADT) utilizando una lista simplemente enlazada que incluye un nodo centinela de cabecera.

```python
class Empty(Exception):                    # error para cuando la cola está vacía
    pass


class LinkedQueue:                         # la clase de la cola
    """Cola FIFO sobre lista enlazada simple con nodo centinela."""

    class _Node:                           # molde para crear nodos
        __slots__ = '_element', '_next'    # cada nodo: dato y gancho al siguiente
        def __init__(self, e, n):
            self._element = e              # guardo el dato
            self._next = n                 # guardo a qué nodo apunta

    def __init__(self):                    # al crear la cola: LinkedQueue()
        self._header = self._Node(None, None)  # centinela: sin dato, sin siguiente
        self._tail = self._header          # cola vacía: el "último" es el centinela
        self._size = 0                     # 0 elementos

    def __len__(self):                     # len(cola)
        return self._size                  # cuántos elementos hay

    def is_empty(self):                    # ¿está vacía?
        return len(self) == 0              # True si hay 0 elementos

    def first(self):                       # mirar el de adelante SIN quitarlo
        if self.is_empty():                # si no hay nadie...
            raise Empty('la cola está vacía')  # ...error
        else:
            return self._header._next._element  # el frente es el nodo justo después del centinela

    def enqueue(self, e):                  # meter e al FINAL de la fila
        nuevo = self._Node(e, None)        # nodo nuevo; apunta a None porque será el último
        self._tail._next = nuevo           # el último actual (o el centinela) engancha al nuevo
        self._tail = nuevo                 # el nuevo pasa a ser el último
        self._size += 1                    # un elemento más

    def dequeue(self):                     # sacar el de ADELANTE y devolverlo
        if self.is_empty():                # si no hay nadie...
            raise Empty('la cola está vacía')  # ...error
        else:
            primero = self._header._next   # guardo el nodo del frente para no perderlo
            self._header._next = primero._next  # el centinela se salta al primero
            self._size -= 1                # un elemento menos
            if self.is_empty():            # si la cola quedó vacía...
                self._tail = self._header  # ..._tail vuelve al centinela
            return primero._element        # devuelvo el dato que saqué


# --- Tests ---
q = LinkedQueue()                          # [header] → None            tail = header
q.enqueue(1)                               # [header] → 1               tail = 1
q.enqueue(2)                               # [header] → 1 → 2           tail = 2
q.enqueue(3)                               # [header] → 1 → 2 → 3       tail = 3

print(len(q), q.first())                   # 3  1   (first solo mira)
print(q.dequeue())                         # 1      queda: [header] → 2 → 3
print(q.dequeue())                         # 2      queda: [header] → 3
print(q.dequeue())                         # 3      queda: [header] → None, tail = header
print(q.is_empty())                        # True
```

Describe un método de tiempo lineal para calcular la longitud de camino de un árbol T.
- Profundidad de un nodo = cuántas líneas hay que bajar desde la raíz para llegar a él.
- Tiempo lineal = O(n): visitar cada nodo una sola vez.

```python
#Ejemplo
            A            ← profundidad 0
          /   \
         B     C         ← profundidad 1
        / \   / \
       D   E F   G       ← profundidad 2

Path length = A + B + C + D + E + F + G
            = 0 + 1 + 1 + 2 + 2 + 2 + 2
            = 10
```

```python
def suma_profundidades(T, node, d):
    if node is None:                             # caso base: no hay nodo
        return 0
    else:
        return (d                                # aporte: la profundidad de este nodo
                + suma_profundidades(T, T.left(node), d + 1)    # hijos: un nivel más abajo
                + suma_profundidades(T, T.right(node), d + 1))

def path_length(T):
    return suma_profundidades(T, T.root(), 0)    # la raíz tiene profundidad 0

# TRAZA con el árbol A-G:
# A (d=0) → 0 + suma(B,1) + suma(C,1)
#   B (d=1) → 1 + suma(D,2) + suma(E,2) = 1 + 2 + 2 = 5
#   C (d=1) → 1 + suma(F,2) + suma(G,2) = 1 + 2 + 2 = 5
# A = 0 + 5 + 5 = 10
```

 - Saber si dos LBT son iguales

Dos árboles son iguales si cumplen las dos cosas a la vez:

Misma forma: los nodos están en los mismos lugares.
Mismos datos: en cada lugar está el mismo elemento.

Errores comunes:
- comparar los nodos con is o == | (Lo que se compara es el dato: n1._element == n2._element)
- comparar solo un recorrido. | (hay que recorrer los dos árboles al mismo tiempo)
- Que un árbol tenga un hijo donde el otro no. | (Si en un lado hay un nodo y en el otro hay None, ya son distintos. Hay que revisar eso antes de leer ._element, para no intentar leer None._element.)

```python
def nodos_iguales(T1, n1, T2, n2):
    if n1 is None and n2 is None:                # 1. los dos lugares vacíos → coinciden
        return True
    elif n1 is None or n2 is None:               # 2. uno vacío y el otro no → distintos
        return False
    elif n1._element != n2._element:             # 3. datos distintos → distintos
        return False
    else:                                        # 4. este lugar coincide: revisar los hijos
        return (nodos_iguales(T1, T1.left(n1), T2, T2.left(n2))      # izquierdo con izquierdo
                and nodos_iguales(T1, T1.right(n1), T2, T2.right(n2)))  # derecho con derecho

def arboles_iguales(T1, T2):
    return nodos_iguales(T1, T1.root(), T2, T2.root())   # arranco por las dos raíces
```

### 7.2 Para la clase: implementa `TablaHashFija`

`TablaHashFija` usa separate chaining como `ChainHashMap` de Goodrich, con dos diferencias:

- **No tiene `_resize`**: la capacidad $N$ se fija al crearla y nunca cambia, aunque $\lambda = n/N$ crezca.
- **Cuenta colisiones** en `self.colisiones`, y trae métodos para medir la tabla: casillas ocupadas y lista más larga.

Cada casilla es una lista de entradas `[clave, valor]`, la misma idea de `UnsortedTableMap`.

`__init__`, `_casilla` y `resumen` ya están dados. Completa cada método marcado con `pass`:

- `__len__()`: número de entradas guardadas.
- `__setitem__(k, v)`: busca `k` **solo** en la lista de la casilla `self._casilla(k)`. Si ya está, reemplaza su valor. Si no está, agrégala al final de esa lista y aumenta `self._n`. Si la lista **ya tenía** alguna entrada antes de agregar, aumenta `self.colisiones`.
- `__getitem__(k)`: retorna el valor de `k` buscando en la lista de su casilla. Si no está, `raise KeyError(k)`.
- `__contains__(k)`: `True` si `k` está en la lista de su casilla.
- `capacidad()`: el número de casillas $N$.
- `casillas_ocupadas()`: cuántas casillas tienen la lista no vacía.
- `lista_mas_larga()`: la longitud de la lista más larga.

**Ojo:** ningún método debe recorrer **toda** la tabla para buscar una clave; eso es lo que la tabla hash evita. Solo `casillas_ocupadas` y `lista_mas_larga` recorren todas las casillas, porque miden la tabla completa.

## Capítulo 7: Lista enlazadas

```python
class SinglyLinkedList:
    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    class _Node:
        __slots__ = '_element', '_next'
        def __init__(self, element, nxt):
            self._element = element
            self._next    = nxt

    def __len__(self):
        return self._size
    
    def is_empty(self):
        return self._size == 0
    
    def first(self):
        return self._head._element

    def insert_first(self, e):
        self._head = self._Node(e, self._head)
        if self._size == 0:
            self._tail = self._head
        self._size += 1

    def insert_last(self, e):
        node = self._Node(e, None)
        if self.is_empty():
            self._head = node
        else:
            self._tail._next = node
        self._tail = node
        self._size += 1

    def remove_first(self):
        val = self._head._element
        self._head = self._head._next
        self._size -= 1
        if self._size == 0:
            self._tail = None
        return val

S = SinglyLinkedList()
for x in ['A', 'B', 'C', 'D']:
    S.insert_last(x)
print(len(S), S.first())    # 4  A
S.remove_first()
print(len(S), S.first())    # 3  B
```

```python
def build(items):
    """Construye una SinglyLinkedList con los elementos dados."""
    L = SinglyLinkedList()
    for x in items:
        L.insert_last(x)
    return L

def to_list(L):
    """Recorre L desde la cabeza y retorna sus elementos en una lista de Python."""
    out, node = [], L._head
    while node is not None:
        out.append(node._element)
        node = node._next
    return out
```

**R-7.2** Describe a good algorithm for concatenating two singly linked lists $L$ and $M$, given only references to the first node of each list, into a single list $L'$ that contains all the nodes of $L$ followed by all the nodes of $M$.

```python
def concatenar(L, M):
    # Caso borde: si L está vacía, L empieza donde empieza M
    if L._head is None:
        L._head = M._head
    else:
        # Caminar hasta el ÚLTIMO nodo de L (el que tiene _next = None)
        actual = L._head
        while actual._next is not None:
            actual = actual._next
        # Enganchar el último de L con el primero de M
        actual._next = M._head

# TRAZA con L: A → B → C,  M: D → E
# actual=A, _next=B → no es None, avanzar
# actual=B, _next=C → no es None, avanzar
# actual=C, _next=None → parar; C._next = D
```

Complejidad: O(n), donde n es el tamaño de L, porque hay que recorrerla para encontrar su último nodo. El tamaño de M no importa: nunca la recorres.

```python
class Empty(Exception):                    # error para cuando la cola está vacía
    pass


class LinkedQueue:                         # la clase de la cola
    """Cola FIFO sobre lista enlazada simple con nodo centinela."""

    def __init__(self):                    # al crear la cola: LinkedQueue()
        self._datos = SinglyLinkedList()               

    def __len__(self):                     # len(cola)
        return self._size                  # cuántos elementos hay

    def is_empty(self):                    # ¿está vacía?
        return len(self) == 0              # True si hay 0 elementos

    def first(self):                       # mirar el de adelante SIN quitarlo
        if self.is_empty():                # si no hay nadie...
            raise Empty('la cola está vacía')  # ...error
        else:
            return self._header._next._element  # el frente es el nodo justo después del centinela

    def enqueue(self, e):
        self._datos.add_last(e)

    def dequeue(self):                     # sacar el de ADELANTE y devolverlo
        if self.is_empty():                # si no hay nadie...
            raise Empty('la cola está vacía')  # ...error
        else:
            return self._datos.remove_first()

    def __str__(self):
        return str(self._datos)


# --- Tests ---
q = LinkedQueue()                          # [header] → None            tail = header
q.enqueue(1)                               # [header] → 1               tail = 1
q.enqueue(2)                               # [header] → 1 → 2           tail = 2
q.enqueue(3)                               # [header] → 1 → 2 → 3       tail = 3

print(len(q), q.first())                   # 3  1   (first solo mira)
print(q.dequeue())                         # 1      queda: [header] → 2 → 3
print(q.dequeue())                         # 2      queda: [header] → 3
print(q.dequeue())                         # 3      queda: [header] → None, tail = header
print(q.is_empty())                        # True
```

**R-9.2** Suppose you label each position $p$ of a binary tree $T$ with a key equal to its preorder rank. Under what circumstances is $T$ a heap?

Suponga que a cada posición p de un árbol binario T le pone como clave su número en preorden. ¿En qué casos T es un heap?

```python
from goodrich.ch06.array_queue import ArrayQueue

def label_preorder(T):
    """Reemplaza el elemento de cada posición por su rango en preorden (0, 1, 2, ...)."""
    for rank, p in enumerate(T.preorder()):
        T._replace(p, rank)
    #Recorre el árbol en preorden y a cada nodo le cambia su valor por su número
    #de visita. enumerate le da a cada nodo su número: 0 al primero, 1 al segundo, etc.

def has_heap_order(T):
    return all(p.element() <= c.element() for p in T.positions() for c in T.children(p))
#Toma cada padre p y cada hijo suyo c, y revisa que p <= c. all(...) da True solo si
# se cumple para todas las parejas. Con números de preorden siempre da True.
def is_complete(T):
    """True si T es un árbol binario completo (BFS: después de un hueco no puede haber nodos)."""
    if T.is_empty():
        return True
    queue, gap = ArrayQueue(), False
    queue.enqueue(T.root())
    while not queue.is_empty():
        p = queue.dequeue()
        for c in (T.left(p), T.right(p)):
            if c is None:
                gap = True
            elif gap:
                return False
            else:
                queue.enqueue(c)
    return True

def preorder_labeled_is_heap(T):
    label_preorder(T)
    return has_heap_order(T) and is_complete(T)

# Respuesta: en preorden un padre siempre se visita antes que sus descendientes, así que
# el orden de heap se cumple SIEMPRE. Por lo tanto T es un heap si y solo si T es completo.
```

```python
from goodrich.ch09.heap_priority_queue import HeapPriorityQueue

class AirTrafficSimulator:
    """Agenda de eventos ordenada por time stamp, implementada con un heap."""

    def __init__(self):
        self._events = HeapPriorityQueue()

    def __len__(self):
        return len(self._events)

    def schedule(self, time, event):
        """Agrega un evento futuro."""
        self._events.add(time, event)          # la hora es la clave (prioridad)

    def next_event(self):
        """Retorna y elimina (time, event) con el menor time stamp."""
        return self._events.remove_min()       # sale el evento más próximo

# Respuesta: se debe usar una cola de prioridad implementada con un heap (clave = time stamp),
# porque insertar (add) y extraer el menor (remove_min) cuestan O(log n) cada una.
# Con una lista sin ordenar, extraer cuesta O(n); con una lista ordenada, insertar cuesta O(n).
```

**C-9.26** Show how to implement the stack ADT using only a priority queue and one additional integer instance variable.

Muestra cómo implementar una pila usando solo una cola de prioridad y una variable entera adicional.”

```python
from goodrich.ch09.heap_priority_queue import HeapPriorityQueue

class PQStack:
    """Pila implementada con una cola de prioridad y un contador entero."""

    def __init__(self):
        self._pq = HeapPriorityQueue()       # la estructura dada
        self._contador = 0                   # la única variable entera extra

    def __len__(self):
        return len(self._pq)

    def is_empty(self):
        return len(self) == 0

    def push(self, e):
        self._contador -= 1                  # cada nuevo tiene clave más pequeña
        self._pq.add(self._contador, e)

    def top(self):
        """Lanza Empty si la pila está vacía."""
        if len(self) == 0:
            raise Empty('La pila está vacía')
        else:
            k, v = self._pq.min()            # mira el más reciente sin sacarlo
            return v

    def pop(self):
        """Lanza Empty si la pila está vacía."""
        if len(self) == 0:
            raise Empty('La pila está vacía')
        else:
            k, v = self._pq.remove_min()     # saca el más reciente
            return v
```

```python
# Pruebas C-9.26
S = PQStack()
assert len(S) == 0 and S.is_empty()
for metodo in (S.top, S.pop):
    try:
        metodo()
        assert False, f'{metodo.__name__} en pila vacía debe lanzar Empty'
    except Empty:
        pass

S.push('a'); S.push('b'); S.push('c')
assert len(S) == 3 and S.top() == 'c'
assert S.pop() == 'c' and S.pop() == 'b'
S.push('d')
assert [S.pop(), S.pop()] == ['d', 'a']
assert S.is_empty()

for i in range(1000):
    S.push(i)
assert [S.pop() for _ in range(1000)] == list(range(999, -1, -1))
print('C-9.26: todas las pruebas pasaron')
```

**C-9.27** Show how to implement the FIFO queue ADT using only a priority queue and one additional integer instance variable.

```python
class PQQueue:
    """Cola FIFO implementada con una cola de prioridad y un contador entero."""


    def __init__(self):
        self._pq = HeapPriorityQueue()       # la estructura dada
        self._contador = 0                   # la única variable entera extra

    def __len__(self):
        return len(self._pq)

    def is_empty(self):
        return len(self) == 0

    def enqueue(self, e):
        self._contador += 1                  # cada nuevo tiene clave más grande
        self._pq.add(self._contador, e)

    def first(self):
        """Lanza Empty si la cola está vacía."""
        if len(self) == 0:
            raise Empty('La cola está vacía')
        else:
            k, v = self._pq.min()            # mira el más antiguo sin sacarlo
            return v

    def dequeue(self):
        """Lanza Empty si la cola está vacía."""
        if len(self) == 0:
            raise Empty('La cola está vacía')
        else:
            k, v = self._pq.remove_min()     # saca el más antiguo
            return v
```

```python
# Pruebas C-9.27
Q = PQQueue()
assert len(Q) == 0 and Q.is_empty()
for metodo in (Q.first, Q.dequeue):
    try:
        metodo()
        assert False, f'{metodo.__name__} en cola vacía debe lanzar Empty'
    except Empty:
        pass

Q.enqueue('a'); Q.enqueue('b'); Q.enqueue('c')
assert len(Q) == 3 and Q.first() == 'a'
assert Q.dequeue() == 'a' and Q.dequeue() == 'b'
Q.enqueue('d')                                  # se intercalan inserciones y extracciones
assert [Q.dequeue(), Q.dequeue()] == ['c', 'd']
assert Q.is_empty()

for i in range(1000):
    Q.enqueue(i)
assert [Q.dequeue() for _ in range(1000)] == list(range(1000))
print('C-9.27: todas las pruebas pasaron')
```

**R-10.9** Draw the 11-entry hash table that results from using the hash function, $h(i) = (3i + 5) \bmod 11$, to hash the keys 12, 44, 13, 88, 23, 94, 11, 39, 20, 16, and 5, assuming collisions are handled by chaining.

“Dibuja la tabla hash de 11 casillas que resulta de usar la función h(i) = (3i + 5) mod 11 para meter las claves 12, 44, 13, 88, 23, 94, 11, 39, 20, 16 y 5. Las colisiones se manejan por encadenamiento (chaining).

```python
N = 11

def hash_chaining(keys):
    """Tabla de N cubetas; cada cubeta es una lista con las claves que caen en ella."""
    tabla = []
    j = 0
    while j < N:
        tabla.append([])                 # crea las N cubetas vacías
        j += 1

    n = len(keys)
    i = 0
    while i < n:
        k = keys[i]
        cubeta = (3 * k + 5) % N         # h(k): en qué cubeta cae
        tabla[cubeta].append(k)          # chaining: se agrega al final de esa lista
        i += 1
    return tabla

T = hash_chaining([12, 44, 13, 88, 23, 94, 11, 39, 20, 16, 5])
# T = [[13], [94, 39], [], [], [], [44, 88, 11], [], [], [12, 23], [16, 5], [20]]
```

```python
# Pruebas R-10.9
table = hash_chaining(KEYS)
assert table == [[13], [94, 39], [], [], [], [44, 88, 11], [], [], [12, 23], [16, 5], [20]]
assert sorted(k for bucket in table for k in bucket) == sorted(KEYS)
assert all(h(k) == j for j, bucket in enumerate(table) for k in bucket)
assert hash_chaining([]) == [[] for _ in range(N)]
print('R-10.9: todas las pruebas pasaron')
```

**R-10.10** What is the result of the previous exercise, assuming collisions are handled by linear probing?

```python
def hash_linear_probing(keys):
    """Tabla de N celdas; si h(k) está ocupada se prueba la siguiente (circularmente)."""
    tabla = [None] * N                   # N celdas vacías
    n = len(keys)
    i = 0
    while i < n:
        k = keys[i]
        j = h(k)                         # casillero donde debería ir
        while tabla[j] is not None:      # mientras esté ocupado...
            j = (j + 1) % N              # ...prueba el siguiente (y vuelve a 0 al final)
        tabla[j] = k                     # encontró uno libre: ahí la guarda
        i += 1
    return tabla
```

```python
# Pruebas R-10.10
table = hash_linear_probing(KEYS)
assert table == [13, 94, 39, 16, 5, 44, 88, 11, 12, 23, 20]
assert sorted(table) == sorted(KEYS)                     # la tabla queda llena
# cada clave está en h(k) o después, y todas las celdas del camino están ocupadas
for j, k in enumerate(table):
    i = h(k)
    while i != j:
        assert table[i] is not None
        i = (i + 1) % N
assert hash_linear_probing([12, 23]) == [None] * 8 + [12, 23, None]
print('R-10.10: todas las pruebas pasaron')
```
